```python
import pandas as pd
from itertools import product

# ----------------- CONFIGURACIÓN -----------------
FILE          = "data/sensibilidad.csv"   # ruta al CSV
TARGET_SAFE   = 0.80    # umbral de éxito mínimo para la zona segura
TARGET_IDEAL  = 0.95    # umbral de éxito mínimo para la zona ideal
# -------------------------------------------------

# 1. Carga y limpieza -------------------------------------------
df = pd.read_csv(FILE)
# elimina cualquier columna cuyo nombre empiece por 'Unnamed'
df = df.loc[:, ~df.columns.str.contains(r"^Unnamed", case=False)]

# 2. Función para buscar la “mejor” zona ------------------------
def find_zone(data, target):
    """
    Devuelve un diccionario con los límites:
         ruido_max, periodo_min, frec_max, success, count
    que maximizan la cobertura manteniendo 'success' ≥ target.
    """
    ruido_vals   = sorted(data["ruido"].unique())
    periodo_vals = sorted(data["periodo"].unique())
    frec_vals    = sorted(data["frec_obs"].unique())

    best = None
    for r_max, p_min, f_max in product(ruido_vals, periodo_vals, frec_vals):
        subset = data[
            (data["ruido"]   <= r_max) &
            (data["periodo"] >= p_min) &
            (data["frec_obs"] <= f_max)]
        if subset.empty:
            continue
        success = subset["ok"].mean()
        if success >= target:
            # prioriza subconjuntos con MÁS filas; si empatan, con ruido_max mayor
            if (best is None or
                len(subset) > best["count"] or
                (len(subset) == best["count"] and r_max > best["ruido_max"])):
                best = dict(
                    ruido_max   = r_max,
                    periodo_min = p_min,
                    frec_max    = f_max,
                    success     = success,
                    count       = len(subset)
                )
    return best

# 3. Cálculo de zonas por cada valor de n_meses ---------------
# Creamos un diccionario con una “tabla” (dos zonas) para cada n_meses
zonas_por_mes = {}
for n in sorted(df["n_meses"].unique()):
    df_n = df[df["n_meses"] == n]
    safe  = find_zone(df_n, TARGET_SAFE)
    ideal = find_zone(df_n, TARGET_IDEAL)
    zonas_por_mes[n] = {"SEGURA": safe, "IDEAL": ideal}

# 4. Presentación -----------------------------------------------
def mostrar(nombre, zona, target):
    if zona is None:
        print(f"{nombre}: ningún rango alcanza el {target*100:.0f}% de éxito.")
        return
    print(f"• ruido    ≤ {zona['ruido_max']}")
    print(f"• periodo  ≥ {zona['periodo_min']}")
    print(f"• frec_obs ≤ {zona['frec_max']}")
    print(f"→ éxito medio: {zona['success']*100:5.2f}%  ({zona['count']} combinaciones)\n")

# Imprimimos 4 “tablas”, una por cada n_meses
for n, zonas in zonas_por_mes.items():
    header = f" n_meses = {n} "
    print(f"\n{header:=^40}")
    print("ZONA SEGURA")
    mostrar("SEGURA", zonas["SEGURA"], TARGET_SAFE)
    print("ZONA IDEAL")
    mostrar("IDEAL",  zonas["IDEAL"],  TARGET_IDEAL)
```
